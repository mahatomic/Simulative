### **Посчитайте Retention 15 дня (в процентах) для пользователей, зарегистрированных в январе**

Cохраните результат в переменную `retention_15_day`

**Примечание:** результат округлите до 5 знаков после запятой

```python
# Ваше решение
def split_lst(path):
  with open(path, 'r') as f:
    data = f.read().split('\n')
    users = []
    date = []
    for line in data[1:]:
      sp_line = line.split(';')
      users.append(sp_line[0])
      date.append(sp_line[1])
  return users, date
```

```python
# Файл с регистрациями
path_reg = '/content/registrations.csv'
result_u, result_d = split_lst(path_reg)
dict_users = dict(zip(result_u, result_d))
print(dict_users)
```

```python
# пользователи из январской когорты
def cohorts(all_users, month):
  users_coh = set([u for u, dt in all_users.items() if dt[:7] == f'2021-{month:02d}'])
  return users_coh

users_j = cohorts(dict_users, 1)
print(users_j)
```

```python
from datetime import datetime
```

```python
# Файл с заходами
path_ent = '/content/entries.csv'
result_un, result_dn = split_lst(path_ent)

# Список с разницей между регистрацией пользователя и каждым его заходом
diffs = []
for u, dt in zip(result_un, result_dn):
    d1 = datetime.strptime(dict_users[u], '%Y-%m-%d')
    d2 = datetime.strptime(dt, '%Y-%m-%d')
    diff = (d2 - d1).days
    diffs.append(diff)

print(diffs)
```

```python
def cur_retention(cohort_users, days, rolling=False):
  users_ndays  = set()
  for u, diff in zip(result_un, diffs):
    if u in cohort_users:
      if rolling:
        match = diff >= days
      else:
        match = diff == days
      if match:
        users_ndays.add(u)
  return users_ndays

users_n15 = cur_retention(users_j, 15)

print(users_n15)
```

```python
def retention(cohort_users, current):
  cohort = len(cohort_users)
  n_cur = len(current)
  retention_cur = n_cur/ cohort
  return retention_cur

retention_15_day = round(retention(users_j, users_n15) * 100, 5)
retention_15_day
```

```python
#@title ✏️ Проверка: чтобы проверить свое решение запустите код в этой ячейке
# Открываем файл с правильными ответами
with open('metrics.txt', 'r') as f:
    answers = f.read().split('\n')

correct_answer = float(answers[0])

try:
    assert retention_15_day == correct_answer
except AssertionError:
    print('Ответы не совпадают')
else:
    print('Поздравляем, Вы справились!')
```

### **Посчитайте Rolling-retention 30 дня (в процентах) для пользователей из той же когорты**

Сохраните результат в переменную `rolling_retention`

**Примечание:** результат округлите до 5 знаков после запятой

```python
# Ваше решение
def cur_retention(cohort_users, days, rolling=False):
  users_ndays  = set()
  for u, diff in zip(result_un, diffs):
    if u in cohort_users:
      if rolling:
        match = diff >= days
      else:
        match = diff == days
      if match:
        users_ndays.add(u)
  return users_ndays

users_n30 = cur_retention(users_j, 30, rolling=True)

print(users_n30)
```

### **Посчитайте Lifetime по всем пользователям, посчитанный как интеграл от n-day retention**

Сохраните результат в переменную `lifetime`

**Примечание:** результат округлите до 5 знаков после запятой

```python
# Ваше решение
lifetime = 0
for i in range(max(diffs) + 1):
  users_n = cur_retention(dict_users, i)
  retention_current = retention(dict_users, users_n)
  lifetime += retention_current

lifetime = round(lifetime, 5)
lifetime
```

### **Посчитайте Churn rate 29 дня (в долях), посчитанный по всем пользователям**

Сохраните результат в переменную `churn_29`.  
Если будете считать CR от Retention, ведите расчет от Rolling Retention, таким образом, мы получим всех, кто не заходил в 29 день и после. Ведя расчет от обычного Retention, наоборот - получим только CR в 29 день.


```python
# Ваше решение
users_29 = cur_retention(dict_users, 29, rolling=True)
retention_29 = retention(dict_users, users_29)

churn_29 = round(1 - retention_29, 5)
churn_29
```
